# 📓 Notebook 02 — Golden Layer (Medalhão) vs Sklearn Pipelines

> **Pergunta operacional:** onde mexer no dado — antes do treino (camada Gold do medalhão) ou dentro do treino (`Pipeline` + validação cruzada)?
>
> **Origem canônica:** [Aula 03, tema 3.1 — Pipelines de pré-processamento](../../../../disciplinas/01-introducao-ao-ciclo-de-vida-de-modelos/temas.md).

## Objetivo

Mostrar a diferença operacional entre **transformações de dado (medalhão)** e **transformações de treino (sklearn Pipeline)**. Saímos com a regra de partição:

| Pertence ao **medalhão (Bronze→Silver→Gold)** | Pertence ao **Pipeline de treino** |
|------------------------------------------------|------------------------------------|
| Padronização de schema, deduplicação, conformação de unidades | Imputação por estatística do fold |
| Junção entre tabelas, materialização de features históricas | One-hot, scaling, encoding por classe |
| Regras de qualidade (quarentena, regras de negócio) | Seleção de features baseada em CV |
| Versionamento e particionamento | Hiperparâmetros do modelo |

## Por que isso importa

Misturar as duas camadas é a causa #1 de **data leakage** em projetos de churn. Se imputo média antes de fazer o split, treino e teste compartilham informação — métrica fica linda e quebra em produção.

## 0. Setup

```python
from __future__ import annotations

from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

RAW_PATH = Path("../dataset/processed/telco_churn.csv")
BRONZE_DIR = Path("../dataset/raw")
SILVER_DIR = Path("../dataset/interim")
GOLD_DIR = Path("../dataset/processed")
for d in (BRONZE_DIR, SILVER_DIR, GOLD_DIR):
    d.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(RAW_PATH)
df.head(3)
```

## 1. Bronze: cópia fiel + quarentena

Bronze é o dado bruto **sem juízo de valor**. Ele responde a duas perguntas: o que chegou? e o que falhou ao chegar? Aqui:

1. salvamos o snapshot original;
2. quarentenamos registros que violam o contrato (do notebook 01).

```python
bronze_path = BRONZE_DIR / "telco_churn_bronze.parquet"
quarantine_path = BRONZE_DIR / "telco_churn_quarantine.parquet"

ok_mask = (
    df["tenure"].between(0, 72)
    & df["monthly_charges"].between(0, 200)
    & df["churn"].isin(["Yes", "No"])
)

bronze = df[ok_mask].copy()
quarantine = df[~ok_mask].copy()
try:
    bronze.to_parquet(bronze_path, index=False)
    quarantine.to_parquet(quarantine_path, index=False)
except Exception:
    bronze.to_csv(bronze_path.with_suffix(".csv"), index=False)
    quarantine.to_csv(quarantine_path.with_suffix(".csv"), index=False)

print(f"Bronze     : {len(bronze):,} linhas")
print(f"Quarentena : {len(quarantine):,} linhas")
```

## 2. Silver: conformação

Silver normaliza unidades, deduplica, padroniza categorias e materializa **features estáveis** que dependem só do registro (não dependem do split de treino).

```python
silver = bronze.drop_duplicates(subset="customer_id").copy()

# Conformação determinística — independente do split.
silver["contract"] = silver["contract"].str.replace("-", " ", regex=False).str.title()
silver["is_senior"] = silver["senior_citizen"].astype(bool)
silver["has_partner"] = silver["partner"].eq("Yes")
silver["has_dependents"] = silver["dependents"].eq("Yes")

# Feature histórica derivada do contrato — não usa estatística do conjunto.
silver["avg_monthly_revenue"] = np.where(
    silver["tenure"] > 0,
    silver["total_charges"] / silver["tenure"].clip(lower=1),
    silver["monthly_charges"],
).round(2)

silver_path = SILVER_DIR / "telco_churn_silver.parquet"
try:
    silver.to_parquet(silver_path, index=False)
except Exception:
    silver.to_csv(silver_path.with_suffix(".csv"), index=False)
print(f"Silver: {len(silver):,} linhas, {silver.shape[1]} colunas")
```

## 3. Gold: tabela de modelagem

Gold é o **artefato consumível pelo treino**. Ele já tem o target separado dos descritivos e descarta colunas operacionais (`customer_id`). Continua **sem nenhuma estatística do conjunto** — quem aplica média, scaling ou encoding é o Pipeline do treino.

```python
TARGET = "churn"
OPERATIONAL = ["customer_id", "senior_citizen", "partner", "dependents"]
gold = silver.drop(columns=OPERATIONAL).copy()
gold[TARGET] = (gold[TARGET] == "Yes").astype(int)

gold_path = GOLD_DIR / "telco_churn_gold.parquet"
try:
    gold.to_parquet(gold_path, index=False)
except Exception:
    gold.to_csv(gold_path.with_suffix(".csv"), index=False)
print("Gold colunas:", list(gold.columns))
```

### 🛑 Breakpoint — discussão

1. Por que `avg_monthly_revenue` pode entrar no Silver, mas `monthly_charges_zscore` **não** pode?
2. Em quais cenários `is_senior` deveria viver no Gold, e quando faz mais sentido só no Pipeline?
3. Quem é o dono de cada camada na sua organização: engenharia de dados, ML, ou produto?

## 4. Sklearn Pipeline — o que SÓ pode acontecer dentro do fold

Imputação, scaling e encoding precisam ser **ajustados apenas no treino** de cada fold. Se vazarem para o teste, a métrica fica enviesada. O `Pipeline` faz isso automaticamente — basta encadear `ColumnTransformer` + estimador.

```python
X = gold.drop(columns=[TARGET])
y = gold[TARGET]

numeric_features = ["tenure", "monthly_charges", "total_charges", "avg_monthly_revenue"]
boolean_features = ["is_senior", "has_partner", "has_dependents"]
categorical_features = [c for c in X.columns if c not in numeric_features + boolean_features]

numeric_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False)),
])
preprocess = ColumnTransformer([
    ("num", numeric_pipe, numeric_features),
    ("cat", categorical_pipe, categorical_features),
    ("bool", "passthrough", boolean_features),
])

model = Pipeline([
    ("prep", preprocess),
    ("clf", LogisticRegression(max_iter=500, random_state=RANDOM_SEED)),
])
model
```

## 5. Validação cruzada estratificada

Churn é desbalanceado. A `StratifiedKFold` preserva a proporção de classes em cada fold. As métricas saem com média e desvio — sem isso, qualquer comparação de modelos é loteria.

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)
scores = cross_validate(
    model,
    X,
    y,
    cv=cv,
    scoring=["roc_auc", "average_precision", "f1"],
    n_jobs=1,
    return_train_score=True,
)

summary = pd.DataFrame(
    {
        "metric": ["roc_auc", "average_precision", "f1"],
        "train_mean": [scores["train_roc_auc"].mean(), scores["train_average_precision"].mean(), scores["train_f1"].mean()],
        "test_mean": [scores["test_roc_auc"].mean(), scores["test_average_precision"].mean(), scores["test_f1"].mean()],
        "test_std": [scores["test_roc_auc"].std(), scores["test_average_precision"].std(), scores["test_f1"].std()],
    }
).round(3)
summary
```

## 6. Anti-padrão: o que NÃO fazer

Para deixar visível o efeito do leakage, vamos imputar e escalar **fora** do Pipeline, depois rodar a mesma CV. A AUC tende a inflar e o desvio cai falsamente.

```python
from copy import deepcopy

X_leak = X.copy()
for col in numeric_features:
    X_leak[col] = (X_leak[col] - X_leak[col].mean()) / X_leak[col].std()
X_leak = pd.get_dummies(X_leak, columns=categorical_features, drop_first=False)

leaky_model = LogisticRegression(max_iter=500, random_state=RANDOM_SEED)
leaky_scores = cross_validate(leaky_model, X_leak, y, cv=cv, scoring=["roc_auc", "average_precision"], n_jobs=1)
print(f"AUC com leakage : {leaky_scores['test_roc_auc'].mean():.3f} ± {leaky_scores['test_roc_auc'].std():.3f}")
print(f"AUC sem leakage : {scores['test_roc_auc'].mean():.3f} ± {scores['test_roc_auc'].std():.3f}")
```

> 🔎 A diferença numérica costuma ser pequena (porque o dataset é estável), mas o **risco aumenta com dados reais**: imputação por média global mascara a heterogeneidade entre regiões/segmentos. A regra fica: **transformação que depende da distribuição → dentro do Pipeline**.
